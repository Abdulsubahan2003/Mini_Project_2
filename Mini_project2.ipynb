```python
cat_cols = df.select_dtypes(include=['category']).columns.tolist()
print('Categorical columns:', cat_cols)


# Function to normalize a categorical column in-place
def normalize_text_column(series, mapping=None, lower=True):
    s = series.astype(str).str.strip()
    if lower:
        s = s.str.lower()
        if mapping:
            s = s.replace(mapping)
            return s.replace({'nan': np.nan}).astype('category')

for col in cat_cols:
    df[col] = normalize_text_column(df[col])
```

```python
from sklearn.impute import SimpleImputer
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(include=['category']).columns.tolist()
print('Numerical cols:', num_cols)
print('Categorical cols:', cat_cols)


num_imputer = SimpleImputer(strategy='median')
if num_cols:
    df[num_cols] = pd.DataFrame(num_imputer.fit_transform(df[num_cols]), columns=num_cols)

for col in cat_cols:
    if df[col].isnull().any():
        try:
            mode_val = df[col].mode().iloc[0]
            df[col].fillna(mode_val, inplace=True)
        except Exception:
            df[col].fillna('Unknown', inplace=True)



print('\nMissing after imputation:')
print(df.isnull().sum())
```

```python
from scipy import stats
def detect_outliers_iqr(series, k=1.5):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - k * iqr
    upper = q3 + k * iqr
    return series[(series < lower) | (series > upper)]




def detect_outliers_zscore(series, threshold=3.0):
    z = np.abs(stats.zscore(series, nan_policy='omit'))
    return series[z > threshold]


# Example: detect outliers in numerical columns and show counts
outlier_summary = {}
for col in num_cols:
    if df[col].nunique() > 2:
        out_iqr = detect_outliers_iqr(df[col])
        out_z = detect_outliers_zscore(df[col])
        outlier_summary[col] = {'iqr_count': len(out_iqr), 'z_count': len(out_z)}


pd.DataFrame(outlier_summary).T.sort_values('iqr_count', ascending=False).head(10)
```

```python
for col in num_cols:
    if df[col].nunique() > 2:
        lower = df[col].quantile(0.01)
        upper = df[col].quantile(0.99)
        df[col] = df[col].clip(lower=lower, upper=upper)
```

```python
target_col = None
for cand in ['Approved', 'approved', 'Approved?', 'approved_flag', 'approved_flag']:
    if cand in df.columns:
        target_col = cand
        break

if target_col is None:
    print('Target column not found. Please set target_col variable to the approval column name.')
else:
    print('Using target column:', target_col)
    if 'gender' in df.columns or 'Gender' in df.columns:
        gcol = 'gender' if 'gender' in df.columns else 'Gender'
        appr = df.groupby(gcol)[target_col].mean().reset_index()
        sns.barplot(x=gcol, y=target_col, data=appr)
        plt.title(f'Approval rate by {gcol}')
        plt.ylabel('Approval rate')
        plt.show()
```
